```python
# creating a function to perform t test on all num columns
from scipy.stats import ttest_ind,f_oneway,chi2_contingency

def perform_ttest(column):
    stay = df[df['Attrition'] == 0][column]
    leave = df[df['Attrition'] == 1][column]
    
    t_stat, p_value = ttest_ind(stay, leave)
    
    print(f"\nColumn: {column}")
    print("T-Statistic:", round(t_stat, 4))
    print("P-Value:",(p_value))
    
    if p_value < 0.05:
        print("Result: Significant difference between groups")
    else:
        print("Result: No significant difference")
        
columns_to_test = [
    'MonthlyIncome',
    'Age',
    'JobSatisfaction',
    'WorkLifeBalance',
    'YearsAtCompany',
    'TotalWorkingYears'
]

for col in columns_to_test:
    perform_ttest(col)
```

#### Anova Test

ANOVA (Analysis of Variance) is a statistical method used to compare means across two or more groups to determine whether at least one group mean is statistically different from the others.

```python
df.columns
```

```python
# Groups for JobLevel vs MonthlyIncome
groups_joblevel = [
    df[df['JobLevel'] == level]['MonthlyIncome']
    for level in df['JobLevel'].unique()
]

# Perform ANOVA
f_stat, p_value = f_oneway(*groups_joblevel)

print("ANOVA Test: JobLevel vs MonthlyIncome")
print("F-Statistic:", round(f_stat, 4))
print("P-Value:", round(p_value, 4))

# Interpretation
if p_value < 0.05:
    print("Result: Significant difference in salary across Job Levels.")
    print("Insight: Higher job levels earn significantly different salaries.")
else:
    print("Result: No significant salary difference across Job Levels.")
```

```python
# Groups for Education
groups_education = [
    df[df['Education'] == edu]['MonthlyIncome']
    for edu in df['Education'].unique()
]

# Perform ANOVA
f_stat, p_value = f_oneway(*groups_education)

print("\nANOVA Test: Education vs MonthlyIncome")
print("F-Statistic:", round(f_stat, 4))
print("P-Value:", round(p_value, 4))

# Interpretation
if p_value < 0.05:
    print("Result: Significant difference in salary across Education levels.")
    print("Insight: Education level affects salary.")
else:
    print("Result: No significant salary difference across Education levels.")
    print("Insight: Education level does not strongly impact salary.")
```

#### Chi Square TEst
The chi-square (χ²) test is a family of nonparametric statistical tests used to examine relationships between categorical variables or to compare observed counts with expected counts under a specified model

```python
# Select dummy categorical columns (contain "_")
categorical_cols = [col for col in df.columns if "_" in col and col != "Attrition"]

for col in categorical_cols:
    print(f"\nChi-Square Test: {col} vs Attrition")
    
    # Create contingency table
    table = pd.crosstab(df[col], df['Attrition'])
    
    # Perform Chi-square test
    chi2, p, dof, expected = chi2_contingency(table)
    
    print("Chi-Square Value:", round(chi2, 4))
    print("P-Value:",(p))
    
    # Interpretation
    if p < 0.05:
        print("Result: Significant relationship with Attrition.")
    else:
        print("Result: No significant relationship with Attrition.")
```

* Accuracy → Overall correct predictions

```python
def evaluate_model(y_test, y_pred, model_name):
    print("Model:", model_name)
    print("Accuracy:", round(accuracy_score(y_test, y_pred), 4))
    print("Precision:", round(precision_score(y_test, y_pred), 4))
    print("Recall:", round(recall_score(y_test, y_pred), 4))
    print("F1 Score:", round(f1_score(y_test, y_pred), 4))
    print("Confusion Matrix:")
    print(confusion_matrix(y_test, y_pred))
    print("-"*40)
```

#### ROC Curve

```python
# function to create ROC Curve 
def plot_roc_curve(y_test, y_probs, model_name):
    fpr, tpr, thresholds = roc_curve(y_test, y_probs)
    auc_score = roc_auc_score(y_test, y_probs)
    
    plt.plot(fpr, tpr, label=f'{model_name} (AUC = {auc_score:.4f})')
    plt.plot([0,1], [0,1], 'k--')  # Diagonal line
    plt.xlabel('False Positive Rate')
    plt.ylabel('True Positive Rate')
    plt.title('ROC Curve')
    plt.legend()
```

```python
# Getting probabalities for model
# Logistic Regression
y_probs_log = log_model.predict_proba(X_test_scaled)[:,1]

# Decision Tree
y_probs_dt = dt_model.predict_proba(X_test)[:,1]

# Random Forest
y_probs_rf = rf_model.predict_proba(X_test)[:,1]

# SVM
y_probs_svm = svm_model.decision_function(X_test_scaled)

# KNN
y_probs_knn = knn_model.predict_proba(X_test_scaled)[:,1]
```

```python
plt.figure(figsize=(10,7))

plot_roc_curve(y_test, y_probs_log, "Logistic Regression")
plot_roc_curve(y_test, y_probs_dt, "Decision Tree")
plot_roc_curve(y_test, y_probs_rf, "Random Forest")
plot_roc_curve(y_test, y_probs_svm, "SVM")
plot_roc_curve(y_test, y_probs_knn, "KNN")

plt.show()
```

* Curves closer to the top-left corner indicate better trade-offs between TPR and FPR. SVM and Logistic Regression curves lie furthest toward the top-left, so they give higher true positive rates for a given false positive rate.
* The Decision Tree curve lies close to the diagonal, so it provides little discrimination (it performs nearly like random guessing).
* KNN and Random Forest are better than random but worse than SVM/Logistic.

```python
#Logistic Regression
param_grid_log = {
    'C': [0.01, 0.1, 1, 10],
    'class_weight': [None, {0:1, 1:5}, {0:1, 1:10}],
    'solver': ['liblinear', 'lbfgs']
}

# Decision Tree
param_grid_dt = {
    'max_depth': [3, 5, 7, None],
    'min_samples_split': [2, 5, 10],
    'class_weight': [None, {0:1, 1:5}, {0:1, 1:10}]
}

# Random Forest
param_grid_rf = {
    'n_estimators': [50, 100, 200],
    'max_depth': [5, 10, None],
    'class_weight': [None, {0:1, 1:5}, {0:1, 1:10}]
}

# SVM
param_grid_svm = {
    'C': [0.1, 1, 10],
    'kernel': ['linear', 'rbf'],
    'class_weight': [None, {0:1, 1:5}, {0:1, 1:10}]
}

# KNN
param_grid_knn = {
    'n_neighbors': [3,5,7,9],
    'weights': ['uniform', 'distance'],
    'metric': ['euclidean', 'manhattan']
}
```

```python
# Function to perform grid search for any model
def grid_search_model(model, param_grid, X_train, y_train):
    grid = GridSearchCV(
        estimator=model,
        param_grid=param_grid,
        scoring='recall',   # focus on Recall
        cv=5,
        verbose=1
    )
    grid.fit(X_train, y_train)
    return grid.best_estimator_, grid.best_params_
```

```python
# Logistic Regression
best_log, best_params_log = grid_search_model(LogisticRegression(max_iter=1000), param_grid_log, X_train_scaled, y_train)

# Decision Tree
best_dt, best_params_dt = grid_search_model(DecisionTreeClassifier(), param_grid_dt, X_train, y_train)

# Random Forest
best_rf, best_params_rf = grid_search_model(RandomForestClassifier(), param_grid_rf, X_train, y_train)

# SVM
best_svm, best_params_svm = grid_search_model(SVC(probability=True), param_grid_svm, X_train_scaled, y_train)

# KNN
best_knn, best_params_knn = grid_search_model(KNeighborsClassifier(), param_grid_knn, X_train_scaled, y_train)
```

```python
def evaluate_best_model(model, X_test, y_test, model_name):
    y_pred = model.predict(X_test)
    print(f"--- {model_name} ---")
    print("Recall:", round(recall_score(y_test, y_pred), 4))
    print("Accuracy:", round(accuracy_score(y_test, y_pred), 4))
    print("F1-Score:", round(f1_score(y_test, y_pred), 4))
    print()
    
# Evaluate all
evaluate_best_model(best_log, X_test_scaled, y_test, "Logistic Regression")
evaluate_best_model(best_dt, X_test, y_test, "Decision Tree")
evaluate_best_model(best_rf, X_test, y_test, "Random Forest")
evaluate_best_model(best_svm, X_test_scaled, y_test, "SVM")
evaluate_best_model(best_knn, X_test_scaled, y_test, "KNN")
```

```python
# List of models
models = [
    ("Logistic Regression", best_log, X_test_scaled),
    ("Decision Tree", best_dt, X_test),
    ("Random Forest", best_rf, X_test),
    ("SVM", best_svm, X_test_scaled),
    ("KNN", best_knn, X_test_scaled)
]

# Store results
results = []

for name, model, X in models:
    y_pred = model.predict(X)
    results.append({
        "Model": name,
        "Recall": recall_score(y_test, y_pred),
        "Accuracy": accuracy_score(y_test, y_pred),
        "F1-Score": f1_score(y_test, y_pred)
    })

# Convert to DataFrame
df_results = pd.DataFrame(results)

# Display all model metrics
print("All Models Metrics:\n", df_results)

# Select the best model based on highest Recall
best_model_row = df_results.loc[df_results['Recall'].idxmax()]
best_model_name = best_model_row['Model']
print("---------------------------------------------")
print("\n     Best Model Based on Recall:")
print(best_model_row)
```
