here we have three error the kb,mb and varies with device 
lets convered the kb,mb into b and remove the m and kb 

```python
#first check if their is any null value
df.isnull().sum()
```

```python
#find the value having k number
df['Size'].loc[df['Size'].str.contains('k')].value_counts().sum()
```

```python
#find the value having mb number
df['Size'].loc[df['Size'].str.contains('M')].value_counts().sum()
```

```python
#find the value having vary with time number
df['Size'].loc[df['Size'].str.contains('Varies with device')].value_counts().sum()
```

here we confirm that the column has no missing values by summing these threevalues and comparing with isnull function

```python
# now convert the whole column into bytes
def convert_size(size):
    if isinstance (size, str):
        if 'k' in size:
            return float(size.replace('k',' '))*1024
        elif 'M' in size:
            return float(size.replace('M'," "))*1024*1024
        elif 'Varies with device' in size:
            return np.nan
        
    return size   
df['Size']=df['Size'].apply(convert_size)
df.rename(columns={'Size': 'size_in_bytes'},inplace=True)
```

```python
df.describe()
```

```python
# now convered b to mb
df['size_in_bytes']=df['size_in_bytes'].apply(lambda x:x/(1024*1024))
```

```python
#lets take look of install columns
df['Installs'].unique
```

```python
df['Installs'].value_counts()
```

```python
df['Installs'].isnull().sum()
```

```python
# remove + sign
df['Installs']=df['Installs'].apply(lambda x: x.replace('+',"") if '+' in str(x) else x)
```

```python
# remove , sign
df['Installs']=df['Installs'].apply(lambda x: x.replace(',','') if ',' in str(x) else x)
```

```python
df['Installs'].value_counts()
```

```python
df.describe()
```

```python

# Function to clean and convert 'Installs' to int
def convert_installs(value):
    if isinstance(value, str):
        # Replace commas and spaces, then convert to int
        value = value.replace(',', '').replace(' ', '')
        try:
            return int(value)
        except ValueError:
            return np.nan  # Return NaN if conversion fails
    return np.nan  # Handle non-string cases

# Apply the function to the 'Installs' column
df['Installs'] = df['Installs'].apply(convert_installs)

print(df)
```
