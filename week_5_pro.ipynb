## Outliers

```python
# separating numerical and categorical variables 

df_num=df.select_dtypes(exclude=object)
df_cat=df.select_dtypes(include=object)
df_num
```

```python
# checking for outliers through boxplot 

plt.figure(figsize=(20,20))

re=1
for i in df_num.columns:
    plt.subplot(3,4,re)
    sns.boxplot(df[i], palette='rainbow')
    plt.title(i)
    re+=1
plt.tight_layout()
```

outliers are present in temperature, apperent temperature, humidity ,pressure and wind speed.the pressure column is skewed due to an outlier.

```python
# removing the outlier by quantile method
for col in df_num:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers=df[(df[col] <= lower_bound) & (df[col] >= upper_bound)]

    print('Outliers: ',outliers)
```

```python
# Removing outliers and making data clean
df = df[(df[col] >= lower_bound) | (df[col] <= upper_bound)]
```

```python
# lets confirm it throug the boxplot
plt.figure(figsize=(20,20))

re=1
for i in df_num.columns:
    plt.subplot(3,4,re)
    sns.boxplot(df[i], palette='rainbow')
    plt.title(i)
    re+=1
plt.tight_layout()
```

from the plots we clraely see that the outliers is removed and points above or lower the bonds in Temperature,apperatent temperatur ,humidity etc are not actually outliers because high wind high humidity and extreme temperature can be real event not noise.
