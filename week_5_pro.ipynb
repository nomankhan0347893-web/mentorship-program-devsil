#### 1.1 Introduction
In today’s data-driven world, understanding weather patterns is crucial for various sectors, including agriculture, transportation, and tourism. This project leverages a historical weather dataset sourced from Kaggle to analyze long-term trends and build predictive models for temperature forecasting. By utilizing statistical methods and machine learning algorithms, the aim is to uncover insightful patterns that can inform decision-making regarding weather-dependent activities.

#### 1.2 Objectives
* Data Exploration: To comprehensively explore the historical weather dataset, identifying key features and understanding their distributions and relationships.
* Data Preprocessing: To clean and prepare the dataset for analysis, ensuring data integrity and accuracy by handling missing values and outliers.
* Predictive Modeling: To develop and train machine learning models that can accurately predict future temperature readings based on historical data.
* Performance Evaluation: To assess the effectiveness of the predictive models using various performance metrics, ensuring the models are reliable and robust.
* Insights and Recommendations: To provide actionable insights based on the analysis and model predictions that can benefit stakeholders in relevant sectors.
  
#### 1.3 Scope
  This project focuses on the analysis of historical temperature data collected from different geographical regions. It will explore relationships among temperature, humidity, wind speed, and precipitation. The scope includes:

* Temporal Analysis: Examining seasonal and annual temperature trends over the years.
* Geospatial Considerations: Analyzing variations in temperature across different locations.
* Modeling Capabilities: Implementing various machine learning algorithms to determine the best-fit model for temperature prediction.
* Limitations: Acknowledging that the dataset may have missing or biased data points, which could impact model performance. Furthermore, the weather phenomena are influenced by numerous external factors, and predictions may not always be accurate.

## Outliers

```python
# separating numerical and categorical variables 

df_num=df.select_dtypes(exclude=object)
df_cat=df.select_dtypes(include=object)
df_num
```

```python
# checking for outliers through boxplot 

plt.figure(figsize=(20,20))

re=1
for i in df_num.columns:
    plt.subplot(3,4,re)
    sns.boxplot(df[i], palette='rainbow')
    plt.title(i)
    re+=1
plt.tight_layout()
```

outliers are present in temperature, apperent temperature, humidity ,pressure and wind speed.the pressure column is skewed due to an outlier.

```python
# removing the outlier by quantile method
for col in df_num:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers=df[(df[col] <= lower_bound) & (df[col] >= upper_bound)]

    print('Outliers: ',outliers)
```

```python
# Removing outliers and making data clean
df = df[(df[col] >= lower_bound) | (df[col] <= upper_bound)]
```

```python
# lets confirm it throug the boxplot
plt.figure(figsize=(20,20))

re=1
for i in df_num.columns:
    plt.subplot(3,4,re)
    sns.boxplot(df[i], palette='rainbow')
    plt.title(i)
    re+=1
plt.tight_layout()
```

from the plots we clraely see that the outliers is removed and points above or lower the bonds in Temperature,apperatent temperatur ,humidity etc are not actually outliers because high wind high humidity and extreme temperature can be real event not noise.


```python
# droping the loud cover from the  dataset b ecause it is an anoumally
df.drop('Loud Cover',axis=1,inplace=True)
```

```python
# 5.1 Training Process

# selecting Features and target
x=df_num.drop("Temperature (C)",axis=1) # features
y=df_num['Temperature (C)'] # Target
```

```python
# apliting data into trainig and splitting
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
# create a dictionary of list of model to evaluate
models={'LinearRegression':LinearRegression(),
       'DecisionTreeRegressor':DecisionTreeRegressor(),
       'RandomForestRegression': RandomForestRegressor(),
     }
```

```python
#train and predict each model with evalyating metric
# creting some empty variable for storing best model score,accuracy and name
best_model=None
best_mse=float('inf')
best_r2=float('inf')
best_mae=float('inf')

for name,model in models.items():
    model.fit(x_train,y_train)
    
    y_pred=model.predict(x_test)

    print(name,' mae: ',mean_absolute_error(y_test,y_pred))
    print(name,' mse ',mean_squared_error(y_test,y_pred))
    print(name,' r2 ',r2_score(y_test,y_pred))

    if mse < best_mse:
        best_model=model
        best_mse=mse
        best_r2=r2
        best_mae=mae




   
```
