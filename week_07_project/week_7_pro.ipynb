#### 5.3.2 Hypothesis Testing

```python
from scipy.stats import ttest_ind

stay = df[df['Attrition'] == 0]['MonthlyIncome']
leave = df[df['Attrition'] == 1]['MonthlyIncome']
t_stat, p_value = ttest_ind(stay, leave)

print("T-Statistic:", t_stat)
print("P-Value:", p_value)

if p_value < 0.05:
    print("There is a significant difference in salary between employees who stayed and left.")
else:
    print("There is no significant salary difference between the two groups.")
```
