    1.1 About Dataset
</h3>

This machine-generated dataset simulates social media engagement data across various metrics, including likes, shares, comments, impressions, sentiment scores, toxicity, and engagement growth. It is designed for analysis and visualization of trends, buzz frequency, public sentiment, and user behavior on digital platforms.

The dataset can be used to:

Identify spikes or drops in engagement

Analyze changes in sentiment over time

Build dashboards for digital trend tracking

Test algorithms for sentiment analysis or trend prediction

<h3 style="color:#CC0000; 
           background-color:#CCFFE5; 
           padding:10px; 
           border-radius:8px;">
    1.2 Feature Description
</h3>

1. IDENTIFIERS & METADATA:
   - post_id: Unique identifier
   - timestamp: When posted (crucial for temporal patterns)
   
2. PLATFORM CHARACTERISTICS:
   - platform: Instagram, Twitter, Facebook, LinkedIn
   - campaign_phase: awareness, consideration, conversion, retention
   
3. CONTENT CHARACTERISTICS:
   - text_content: The actual post text
   - hashtags: #tags used
   - mentions: @mentions included
   - media_type: image, video, carousel, text-only
   
4. AUDIENCE CONTEXT:
   - location: Geographic targeting
   - language: Content language
   - follower_count: Account size at posting time

<h3 style="color:#CC0000; 
           background-color:#CCFFE5; 
           padding:10px; 
           border-radius:8px;">
    1.3 Target Variable
</h3>

 What we want to predict:

→ engagement_rate: The percentage of viewers who interact with content.\
  Formula: (likes + shares + comments) / impressions × 100
  
  Why this matters: 
  - Raw likes can be misleading (more followers = more likes)
  - Engagement rate normalizes for reach
  - Industry benchmark: 1-3% is average, 6%+ is excellent

```python
# Calculate missing statistics
missing_data = df.isnull().sum()
missing_percent = (missing_data / len(df) * 100)
missing_table = pd.DataFrame({
    'Missing_Count': missing_data,
    'Missing_Percentage': missing_percent.round(2),
    'Data_Type': df.dtypes
}).sort_values('Missing_Percentage', ascending=False)

significant_missing = missing_table[missing_table['Missing_Count'] > 0]
print("🔍 MISSING DATA BREAKDOWN:")
print(significant_missing if len(significant_missing) > 0 else "✅ No missing values found!")

# Visualize missing data pattern
if len(significant_missing) > 0:
    plt.figure(figsize=(12, 6))
    significant_missing_plot = significant_missing.head(10)
    bars = plt.bar(significant_missing_plot.index, significant_missing_plot['Missing_Percentage'], 
                   color=['red' if x > 20 else 'orange' if x > 5 else 'yellow' for x in significant_missing_plot['Missing_Percentage']])
    
    plt.title('Missing Data by Column (%)', fontsize=14, fontweight='bold')
    plt.xlabel('Columns')
    plt.ylabel('Missing Percentage')
    plt.xticks(rotation=45, ha='right')
    
    # Add percentage labels
    for bar in bars:
        height = bar.get_height()
        plt.text(bar.get_x() + bar.get_width()/2., height,
                f'{height:.1f}%', ha='center', va='bottom')
    
    plt.axhline(y=5, color='red', linestyle='--', alpha=0.7, label='5% Threshold')
    plt.axhline(y=20, color='darkred', linestyle='--', alpha=0.7, label='20% Threshold')
    plt.legend()
    plt.tight_layout()
    plt.show()
    
    print("\n⚠️  MISSING DATA STRATEGY:")
    print("   • <5% missing: Listwise deletion acceptable")
    print("   • 5-20% missing: Imputation recommended")
    print("   • >20% missing: Consider dropping feature or advanced imputation")
```

```python
# Check for exact duplicates
exact_duplicates = df.duplicated().sum()
print(f"🔍 Exact duplicate rows: {exact_duplicates:,} ({exact_duplicates/len(df)*100:.2f}%)")

# Check for ID-based duplicates (same post_id)
if 'post_id' in df.columns:
    id_duplicates = df['post_id'].duplicated().sum()
    print(f"🆔 Duplicate post_ids: {id_duplicates:,}")

# Check for content-based duplicates (same text, different ID)
if 'text_content' in df.columns:
    text_duplicates = df['text_content'].duplicated().sum()
    print(f"📝 Duplicate text content: {text_duplicates:,}")

if exact_duplicates > 0:
    print("\n📋 SAMPLE DUPLICATES:")
    duplicates = df[df.duplicated(keep=False)].sort_values(list(df.columns))
    display(duplicates.head(6))
    
    print("\n💡 RECOMMENDATION: Remove exact duplicates to prevent data leakage")
else:
    print("\n✅ No duplicates found. Data integrity maintained!")
```

<h3 style="color:#CC0000; 
           background-color:#CCFFE5; 
           padding:10px; 
           border-radius:8px;">
    2.4 Cradinality Assessment
</h3>

```python
# create cat an num variables
categorical_cols = df.select_dtypes(include=['object', 'category']).columns
numerical_cols = df.select_dtypes(include=['int', 'float']).columns 
```

```python
cardinality_df = pd.DataFrame({
    'Column': df.columns,
    'Unique_Count': [df[col].nunique() for col in df.columns],
    'Cardinality_Type': ['High' if df[col].nunique() > 100 else 
                        'Medium' if df[col].nunique() > 10 else 'Low' 
                        for col in df.columns]
})

print(cardinality_df.sort_values('Unique_Count', ascending=False).to_string(index=False))

# Identify high cardinality categorical columns (potential issues)
high_card_cat = [col for col in categorical_cols if df[col].nunique() > 50]
print(f"\n⚠️  HIGH CARDINALITY CATEGORICAL COLUMNS (>50 unique values):")
for col in high_card_cat:
    print(f"   • {col}: {df[col].nunique()} unique values")
    print(f"     Examples: {df[col].unique()[:5]}")

print(f"\n💡 INSIGHT: High cardinality features may need:")
print("   • Target encoding for ML models")
print("   • Grouping rare categories")
print("   • Feature hashing for text data")
```

<h3 style="color:#CC0000; 
           background-color:#CCFFE5; 
           padding:10px; 
           border-radius:8px;">
    2.5 Statistical Summary
</h3>

```python
if len(numerical_cols) > 0:
    print("🔢 NUMERICAL FEATURES SUMMARY:")
    stats_df = df[numerical_cols].describe().T
    
    # Add advanced statistics
    stats_df['skewness'] = df[numerical_cols].skew()
    stats_df['kurtosis'] = df[numerical_cols].kurtosis()
    stats_df['missing'] = df[numerical_cols].isnull().sum()
    stats_df['zeros'] = (df[numerical_cols] == 0).sum()
    
    print(stats_df.round(2))
    
    print("\n📈 DISTRIBUTION INSIGHTS:")
    for col in numerical_cols[:5]:  # Top 5 columns
        skew = df[col].skew()
        kurt = df[col].kurtosis()
        
        if abs(skew) > 2:
            shape = "highly skewed (viral outliers likely)"
        elif abs(skew) > 0.5:
            shape = "moderately skewed"
        else:
            shape = "approximately normal"
            
        print(f"   • {col}: {shape} (skew={skew:.2f}, kurt={kurt:.2f})")

if len(categorical_cols) > 0:
    print(f"\n🏷️  CATEGORICAL FEATURES SUMMARY:")
    for col in categorical_cols[:5]:  # Show top 5
        print(f"\n   {col.upper()} (Top 5 values):")
        print(df[col].value_counts().head().to_string().replace('\n', '\n      '))
```

</h3>

```python
outlier_report = []

for col in ['likes_count', 'shares_count', 'comments_count', 'impressions', 'engagement_rate']:
    if col not in df.columns:
        continue
        
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    
    outliers = df[(df[col] < lower_bound) | (df[col] > upper_bound)]
    outlier_percent = len(outliers) / len(df) * 100
    
    outlier_report.append({
        'Column': col,
        'Outliers': len(outliers),
        'Percentage': f"{outlier_percent:.2f}%",
        'Treatment': 'Capped (Winsorization)'
    })
    
    # Winsorization (cap at bounds rather than remove)
    df[col] = df[col].clip(lower=lower_bound, upper=upper_bound)

outlier_df = pd.DataFrame(outlier_report)
print("📊 OUTLIER TREATMENT REPORT:")
print(outlier_df.to_string(index=False))

print("\n💡 RATIONALE: Used Winsorization (capping) instead of deletion")
print("   to preserve data volume while reducing extreme value impact.")
```

```python
if 'sentiment_score' in df.columns and 'engagement_rate' in df.columns:
    # Cross-tabulation
    sentiment_engagement = df.groupby('sentiment_score')['engagement_rate'].agg([
        'count', 'mean', 'median', 'std'
    ]).round(2)
    
    print("😊 SENTIMENT PERFORMANCE:")
    print(sentiment_engagement.sort_values('mean', ascending=False))
    
    # Chi-square test for independence [^23^]
    # Create engagement categories for chi-square
    df['engagement_category'] = pd.cut(df['engagement_rate'], 
                                      bins=[0, 1, 3, 6, 100], 
                                      labels=['Low', 'Medium', 'High', 'Viral'])
    
    contingency_table = pd.crosstab(df['sentiment_score'], df['engagement_category'])
    chi2, p_val, dof, expected = chi2_contingency(contingency_table)
    
    print(f"\n🔗 CHI-SQUARE TEST (Sentiment vs Engagement Category):")
    print(f"   Chi-square: {chi2:.2f}")
    print(f"   p-value: {p_val:.2e}")
    print(f"   Association: {'Significant' if p_val < 0.05 else 'Not significant'}")
    
    # Visualization
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(15, 6))
    
    # Bar plot
    sentiment_order = sentiment_engagement.sort_values('mean', ascending=False).index
    sns.barplot(x=sentiment_order, y='mean', data=sentiment_engagement.reset_index(), 
                palette=['green', 'gray', 'red'], ax=ax1)
    ax1.set_title('Average Engagement by Sentiment', fontweight='bold')
    ax1.set_ylabel('Engagement Rate (%)')
    
    # Heatmap of contingency table
    sns.heatmap(contingency_table, annot=True, fmt='d', cmap='YlOrRd', ax=ax2)
    ax2.set_title('Sentiment vs Engagement Category Heatmap', fontweight='bold')
    
    plt.tight_layout()
    plt.show()
```

```python
# 8.1 Engagement Rate (Target Variable)
if all(col in df.columns for col in ['likes_count', 'shares_count', 'comments_count', 'impressions']):
    df['total_engagement'] = df['likes_count'] + df['shares_count'] + df['comments_count']
    df['engagement_rate'] = (df['total_engagement'] / df['impressions']) * 100
    df['engagement_rate'] = df['engagement_rate'].clip(0, 100)  # Cap at 100%
    print("✅ Target variable created: engagement_rate")

# 8.2 Engagement Composition Features
if all(col in df.columns for col in ['likes_count', 'shares_count', 'comments_count']):
    df['like_ratio'] = df['likes_count'] / (df['total_engagement'] + 1)
    df['share_ratio'] = df['shares_count'] / (df['total_engagement'] + 1)
    df['comment_ratio'] = df['comments_count'] / (df['total_engagement'] + 1)
    print("✅ Engagement composition ratios created")

# 8.3 Text Features
if 'text_content' in df.columns:
    df['text_length'] = df['text_content'].str.len()
    df['word_count'] = df['text_content'].str.split().str.len()
    df['avg_word_length'] = df['text_content'].str.replace(' ', '').str.len() / (df['word_count'] + 1)
    print("✅ Text features created")

if 'hashtags' in df.columns:
    df['hashtag_count'] = df['hashtags'].str.count('#')
    df['hashtag_density'] = df['hashtag_count'] / (df['word_count'] + 1)
    print("✅ Hashtag features created")

# 8.4 Time Features (already created in Phase 4, summary here)
time_features = ['hour', 'day_of_week', 'month', 'is_weekend', 'is_business_hours']
existing_time = [f for f in time_features if f in df.columns]
print(f"✅ Time features available: {', '.join(existing_time)}")

# 8.5 Interaction Features
if 'follower_count' in df.columns and 'engagement_rate' in df.columns:
    df['engagement_per_follower'] = df['total_engagement'] / (df['follower_count'] + 1)
    print("✅ Interaction feature created: engagement_per_follower")

print(f"\n📊 TOTAL FEATURES AVAILABLE: {len(df.columns)}")
print("🏗️  Feature engineering foundation complete!")
```

           border-radius:8px;">
   8.2 Hypothesis Testing
</h3>

```python
# Test 1: T-Test - Weekend vs Weekday engagement
if 'is_weekend' in df.columns and 'engagement_rate' in df.columns:
    weekday_engagement = df[df['is_weekend'] == 0]['engagement_rate']
    weekend_engagement = df[df['is_weekend'] == 1]['engagement_rate']
    
    t_stat, p_val = stats.ttest_ind(weekday_engagement, weekend_engagement)
    
    print("📊 TEST 1: Weekend vs Weekday Engagement (T-Test)")
    print(f"   H0: No difference between weekend and weekday engagement")
    print(f"   T-statistic: {t_stat:.3f}")
    print(f"   p-value: {p_val:.2e}")
    print(f"   Result: {'Reject H0' if p_val < 0.05 else 'Fail to reject H0'}")
    print(f"   Interpretation: {'Significant difference' if p_val < 0.05 else 'No significant difference'}")
    print()
```

```python
# Test 2: ANOVA - Platform differences [^23^]
if 'platform' in df.columns:
    platforms = df['platform'].unique()
    if len(platforms) >= 3:
        platform_groups = [df[df['platform'] == p]['engagement_rate'] for p in platforms]
        f_stat, p_val = f_oneway(*platform_groups)
        
        print("📊 TEST 2: Platform Differences (ANOVA)")
        print(f"   H0: All platforms have equal mean engagement")
        print(f"   F-statistic: {f_stat:.3f}")
        print(f"   p-value: {p_val:.2e}")
        print(f"   Result: {'Reject H0' if p_val < 0.05 else 'Fail to reject H0'}")
        print(f"   Interpretation: {'At least one platform differs significantly' if p_val < 0.05 else 'No significant differences'}")
        print()
```

```python
# Test 3: Chi-Square - Sentiment vs Engagement Level [^23^]
if 'sentiment_score' in df.columns:
    contingency = pd.crosstab(df['sentiment_score'], df['engagement_category'])
    chi2, p_val, dof, expected = chi2_contingency(contingency)
    
    print("📊 TEST 3: Sentiment vs Engagement Category (Chi-Square)")
    print(f"   H0: Sentiment and engagement level are independent")
    print(f"   Chi-square: {chi2:.3f}")
    print(f"   p-value: {p_val:.2e}")
    print(f"   Result: {'Reject H0' if p_val < 0.05 else 'Fail to reject H0'}")
    print(f"   Interpretation: {'Significant association' if p_val < 0.05 else 'No significant association'}")

print("\n✅ Statistical testing complete! Insights now have mathematical confidence.")
```

We've observed patterns, but are they real or random luck? Hypothesis testing is our "court of law" where we prove our case. The p-value is our evidence strength—below 0.05 means we're 95% confident the pattern is genuine. T-tests compare two groups, ANOVA compares three or more, Chi-Square tests relationships between categories.

```python
# Select features for modeling
feature_columns = [
    # Engagement components
    'likes_count', 'shares_count', 'comments_count', 'impressions',
    
    # Text features
    'text_length', 'word_count', 'hashtag_count', 'mention_count',
    
    # Time features
    'hour', 'is_weekend', 'is_business_hours',
    
    # Derived ratios
    'like_ratio', 'share_ratio', 'comment_ratio',
    
    # Categorical encodings (if available)
    'sentiment_encoded' if 'sentiment_encoded' in df.columns else None,
    'platform_encoded' if 'platform_encoded' in df.columns else None,
]

# Remove None values
feature_columns = [f for f in feature_columns if f is not None and f in df.columns]

# Create feature matrix X and target y
X = df[feature_columns].copy()
y = df['engagement_rate'].copy()

# Handle any remaining missing values
X = X.fillna(X.median())

print(f"✅ Feature matrix created:")
print(f"   Features selected: {len(feature_columns)}")
print(f"   Samples: {X.shape[0]:,}")
print(f"\n📋 FEATURE LIST:")
for i, feat in enumerate(feature_columns, 1):
    print(f"   {i:2d}. {feat}")

# Encode categorical variables if not already done
if 'platform' in df.columns and 'platform_encoded' not in df.columns:
    le = LabelEncoder()
    X['platform_encoded'] = le.fit_transform(df['platform'])
    feature_columns.append('platform_encoded')
    print(f"\n🔢 Platform encoded: {dict(zip(le.classes_, le.transform(le.classes_)))}")

if 'sentiment_score' in df.columns and 'sentiment_encoded' not in df.columns:
    sentiment_map = {'positive': 2, 'neutral': 1, 'negative': 0}
    X['sentiment_encoded'] = df['sentiment_score'].map(sentiment_map).fillna(1)
    feature_columns.append('sentiment_encoded')
    print(f"🔢 Sentiment encoded: {sentiment_map}")
```

   9.11 HYPERPARAMETER TUNING

```python
from sklearn.model_selection import RandomizedSearchCV  # Faster than GridSearchCV

# Only tune if best model benefits from it
if 'Random Forest' in best_model_name:
    print("🔧 Tuning Random Forest (Randomized Search - Faster)...")
    
    # Smaller, smarter parameter space
    param_distributions = {
        'n_estimators': [50, 100, 150],  # Reduced options
        'max_depth': [10, 15, 20],       # Focused range
        'min_samples_split': [2, 5, 10],
        'min_samples_leaf': [1, 2, 4]
    }
    
    base_model = RandomForestRegressor(random_state=42, n_jobs=-1)
    
    # RandomizedSearchCV: Tests 20 random combinations instead of all
    random_search = RandomizedSearchCV(
        base_model,
        param_distributions,
        n_iter=20,              # Only 20 combinations (not all)
        cv=3,                   # 3-fold CV (faster than 5)
        scoring='r2',
        n_jobs=-1,
        random_state=42,
        verbose=1
    )
    
    print("⏳ Training (this will take 1-3 minutes)...")
    random_search.fit(X_train_unscaled, y_train)
    
    print(f"\n✅ BEST PARAMETERS FOUND:")
    for param, value in random_search.best_params_.items():
        print(f"   {param}: {value}")
    
    print(f"\n📊 Cross-Validation R²: {random_search.best_score_:.4f}")
    
    # Evaluate on test set
    best_tuned_model = random_search.best_estimator_
    y_pred_tuned = best_tuned_model.predict(X_test_unscaled)
    
    tuned_r2 = r2_score(y_test, y_pred_tuned)
    tuned_rmse = np.sqrt(mean_squared_error(y_test, y_pred_tuned))
    
    print(f"\n🎯 TUNED MODEL TEST PERFORMANCE:")
    print(f"   R²: {tuned_r2:.4f} (vs {results_df.iloc[0]['R²']:.4f} before)")
    print(f"   RMSE: {tuned_rmse:.4f} (vs {results_df.iloc[0]['RMSE']:.4f} before)")
    
    improvement = ((tuned_r2 - results_df.iloc[0]['R²']) / results_df.iloc[0]['R²'] * 100) if results_df.iloc[0]['R²'] > 0 else 0
    print(f"\n📈 Improvement: {improvement:+.1f}% in R² score")

elif 'SVM' in best_model_name:
    print("🔧 Tuning SVM (Limited Search - SVM is slow)...")
    
    # Very limited search for SVM (it's computationally expensive)
    param_distributions = {
        'C': [0.1, 1, 10],
        'gamma': ['scale', 'auto']
    }
    
    base_model = SVR(kernel='rbf')
    
    random_search = RandomizedSearchCV(
        base_model,
        param_distributions,
        n_iter=6,               # Very limited (C × gamma combinations)
        cv=3,
        scoring='r2',
        n_jobs=-1,
        random_state=42
    )
    
    print("⏳ Training SVM (this may take 2-5 minutes)...")
    random_search.fit(X_train_scaled, y_train)
    
    print(f"\n✅ Best Parameters: {random_search.best_params_}")
    best_tuned_model = random_search.best_estimator_
    y_pred_tuned = best_tuned_model.predict(X_test_scaled)
    
    tuned_r2 = r2_score(y_test, y_pred_tuned)
    tuned_rmse = np.sqrt(mean_squared_error(y_test, y_pred_tuned))
    
    print(f"Tuned R²: {tuned_r2:.4f}")

else:
    print("✅ Best model is simple - skipping extensive tuning")
    print("   Using default parameters (already optimal for this model)")
    best_tuned_model = final_model
    y_pred_tuned = final_predictions
    tuned_r2 = final_r2
    tuned_rmse = final_rmse

print("\n✅ Hyperparameter tuning complete!")
```
